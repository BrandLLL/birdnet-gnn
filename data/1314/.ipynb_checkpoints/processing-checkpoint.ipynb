```python
from sklearn.neighbors import BallTree
import numpy as np

# Load station inventory filtered to US + your date range
stations = pd.read_csv('isd-history.csv', dtype=str)
us = stations[
    (stations['ST'].notna()) &
    (stations['BEGIN'] <= '20130101') &
    (stations['END'] >= '20141231') &
    (stations['LAT'].notna()) &
    (stations['LON'].notna())
].copy()
us['LAT'] = us['LAT'].astype(float)
us['LON'] = us['LON'].astype(float)
us['filename'] = us['USAF'] + us['WBAN'] + '.csv'

# Build BallTree on LCD stations
station_coords = np.radians(us[['LAT', 'LON']].values)
tree = BallTree(station_coords, metric='haversine')

# Match ZIP centroids to nearest station
zip_coords = np.radians(zip_centroids[['LAT', 'LON']].values)
distances, indices = tree.query(zip_coords, k=1)

zip_centroids['nearest_station_file'] = us.iloc[indices[:, 0]]['filename'].values
zip_centroids['nearest_station_name'] = us.iloc[indices[:, 0]]['STATION NAME'].values
zip_centroids['nearest_station_km'] = distances[:, 0] * 6371

# Join back to NRSA sites
sites_with_zip = sites_with_zip.merge(
    zip_centroids[['ZIP', 'nearest_station_file', 'nearest_station_name', 'nearest_station_km']],
    on='ZIP', how='left'
)
```

```python
import urllib.request, os

years = ['2013', '2014']
base_url = 'https://www.ncei.noaa.gov/data/local-climatological-data/access/'

# Only unique stations needed — this is the reduction benefit
needed_files = sites_with_zip['nearest_station_file'].dropna().unique()
print(f"Unique stations to download: {len(needed_files)}")  # Much less than 1,900

for year in years:
    os.makedirs(f'lcd/{year}', exist_ok=True)
    for fname in needed_files:
        url = f'{base_url}{year}/{fname}'
        out = f'lcd/{year}/{fname}'
        if not os.path.exists(out):
            try:
                urllib.request.urlretrieve(url, out)
                print(f'Downloaded: {fname} {year}')
            except Exception as e:
                print(f'Missing: {fname} {year} — {e}')
```

```python
TARGET_CHARACTERISTICS = [
    # Oxygen & organics (EPT most sensitive)
    "Dissolved oxygen",
    "Dissolved oxygen saturation",
    "Biochemical oxygen demand",
    
    # Nutrients (eutrophication stress)
    "Nitrogen",
    "Nitrate",
    "Phosphorus",
    "Total dissolved solids",
    
    # Acidity / general chemistry
    "pH",
    "Specific conductance",
    "Turbidity",
    
    # Temperature (thermal regime)
    "Temperature, water",
]
```

```python
import requests
import pandas as pd
import io
import time

def query_wqp_stations(bbox, start_date, end_date, site_type="Stream"):
    """
    bbox: (min_lon, min_lat, max_lon, max_lat)
    dates: "MM-DD-YYYY" strings
    Returns a DataFrame of WQP monitoring stations.
    """
    url = "https://www.waterqualitydata.us/data/Station/search"
    params = {
        "bBox": f"{bbox[0]},{bbox[1]},{bbox[2]},{bbox[3]}",
        "startDateLo": start_date,
        "startDateHi": end_date,
        "siteType": site_type,
        "sampleMedia": "Water",
        "mimeType": "csv",
    }
    resp = requests.get(url, params=params, timeout=120)
    resp.raise_for_status()
    return pd.read_csv(io.StringIO(resp.text), dtype=str)
```

```python
def make_bbox(lats, lons, buffer_deg=0.5):
    return (
        lons.min() - buffer_deg,
        lats.min() - buffer_deg,
        lons.max() + buffer_deg,
        lats.max() + buffer_deg
    )

# Group NRSA sites by state and query WQP stations per state
all_wqp_stations = []

for state, group in nrsa_sites.groupby('STATE'):
    bbox = make_bbox(group['LAT_DD83'], group['LON_DD83'])
    try:
        stations = query_wqp_stations(
            bbox=bbox,
            start_date="01-01-2013",
            end_date="12-31-2014"
        )
        stations['query_state'] = state
        all_wqp_stations.append(stations)
        time.sleep(1)  # be polite to the API
    except Exception as e:
        print(f"Failed for state {state}: {e}")

wqp_stations = pd.concat(all_wqp_stations, ignore_index=True)
wqp_stations['lat'] = pd.to_numeric(wqp_stations['LatitudeMeasure'], errors='coerce')
wqp_stations['lon'] = pd.to_numeric(wqp_stations['LongitudeMeasure'], errors='coerce')
wqp_stations = wqp_stations.dropna(subset=['lat', 'lon'])
```

```python
from sklearn.neighbors import BallTree
import numpy as np

# Build spatial index on WQP stations
wqp_coords = np.radians(wqp_stations[['lat', 'lon']].values)
tree = BallTree(wqp_coords, metric='haversine')

# Match each NRSA site to nearest WQP station
nrsa_coords = np.radians(nrsa_sites[['LAT_DD83', 'LON_DD83']].values)
distances, indices = tree.query(nrsa_coords, k=1)

nrsa_sites['wqp_station_id'] = wqp_stations.iloc[indices[:, 0]]['MonitoringLocationIdentifier'].values
nrsa_sites['wqp_station_km'] = distances[:, 0] * 6371

# Flag matches that are too far — WQP station > 5 km away is a different stream
nrsa_sites['wqp_match_reliable'] = nrsa_sites['wqp_station_km'] <= 5.0

print(f"Sites with reliable WQP match: {nrsa_sites['wqp_match_reliable'].sum()} / {len(nrsa_sites)}")
```

```python
import requests, zipfile, io

def query_wqp_results_post(site_ids, characteristics, start_date, end_date):
    """
    Downloads chemistry results via POST for a list of site IDs.
    Returns a DataFrame.
    """
    url = "https://www.waterqualitydata.us/data/Result/search?mimeType=csv&zip=yes"
    payload = {
        "siteid": site_ids,
        "characteristicName": characteristics,
        "startDateLo": start_date,
        "startDateHi": end_date,
        "sampleMedia": ["Water"],
        "siteType": ["Stream"],
        "dataProfile": "resultPhysChem"
    }
    resp = requests.post(
        url,
        json=payload,
        headers={"Content-Type": "application/json", "Accept": "application/zip"},
        timeout=300
    )
    resp.raise_for_status()
    
    # Unzip the response
    z = zipfile.ZipFile(io.BytesIO(resp.content))
    csv_name = [n for n in z.namelist() if n.endswith('.csv')][0]
    return pd.read_csv(z.open(csv_name), dtype=str)

# Only query sites with reliable matches, in batches of 300 (API limit)
reliable_sites = nrsa_sites[nrsa_sites['wqp_match_reliable']]['wqp_station_id'].unique().tolist()

BATCH_SIZE = 300
all_results = []

for i in range(0, len(reliable_sites), BATCH_SIZE):
    batch = reliable_sites[i:i + BATCH_SIZE]
    try:
        results = query_wqp_results_post(
            site_ids=batch,
            characteristics=TARGET_CHARACTERISTICS,
            start_date="01-01-2012",   # one year buffer before NRSA sampling
            end_date="12-31-2014"
        )
        all_results.append(results)
        print(f"Batch {i//BATCH_SIZE + 1}: {len(results)} records")
        time.sleep(2)
    except Exception as e:
        print(f"Batch {i//BATCH_SIZE + 1} failed: {e}")

wqp_results = pd.concat(all_results, ignore_index=True)
```
